### Separate Red, Green & Blue Channels

```python
# Read the image
Image = cv.imread("Image.png")

if Image is not None:

    # 1. Convert BGR to RGB
    RGB = cv.cvtColor(Image, cv.COLOR_BGR2RGB)

    # 2. Convert image to grayscale
    Gray = cv.cvtColor(Image, cv.COLOR_BGR2GRAY)

    # 3. Separate RGB channels
    Red = RGB[:, :, 0]
    Green = RGB[:, :, 1]
    Blue = RGB[:, :, 2]

    # 4. Create a 2 x 3 figure
    plt.figure(figsize=(12, 8))

    # RGB image
    plt.subplot(2, 3, 1)
    plt.imshow(RGB)
    plt.title("RGB Image")
    plt.axis("off")

    # Red channel
    plt.subplot(2, 3, 2)
    plt.imshow(Red, cmap="gray")
    plt.title("Red Channel")
    plt.axis("off")

    # Green channel
    plt.subplot(2, 3, 3)
    plt.imshow(Green, cmap="gray")
    plt.title("Green Channel")
    plt.axis("off")

    # Blue channel
    plt.subplot(2, 3, 4)
    plt.imshow(Blue, cmap="gray")
    plt.title("Blue Channel")
    plt.axis("off")

    # Grayscale image
    plt.subplot(2, 3, 5)
    plt.imshow(Gray, cmap="gray")
    plt.title("Grayscale Image")
    plt.axis("off")

    # Original BGR image displayed without conversion
    plt.subplot(2, 3, 6)
    plt.imshow(Image)
    plt.title("Original BGR Image")
    plt.axis("off")

    # Display all images
    plt.tight_layout()
    plt.show()

    # 5. Get center pixel RGB value
    height, width = RGB.shape[:2]

    center_row = height // 2
    center_col = width // 2

    center_pixel = RGB[center_row, center_col]

    # Extract RGB values
    R = int(center_pixel[0])
    G = int(center_pixel[1])
    B = int(center_pixel[2])

    # Convert RGB to HEX
    HEX = "#{:02X}{:02X}{:02X}".format(R, G, B)

    # Print center pixel information
    print("Center Pixel RGB:", (R, G, B))
    print("Center Pixel HEX:", HEX)

else:
    print("Error: Image not found!")
```

```python
gradient = np.tile(np.arange(256, dtype=np.uint8), (100, 1))   # 8-bit gradient, 0 to 255
 
def quantize(img, bits):
    levels = 2 ** bits
    step = 256 // levels
    return ((img // step) * (255 // (levels - 1))).astype(np.uint8)
 
gradients = [("Original 8-bit (256 levels)", gradient),
             ("1-bit (2 levels)", quantize(gradient, 1)),
             ("2-bit (4 levels)", quantize(gradient, 2)),
             ("4-bit (16 levels)", quantize(gradient, 4))]
 
fig, axes = plt.subplots(4, 1, figsize=(8, 6))
for ax, (title, img) in zip(axes, gradients):
    ax.imshow(img, cmap="gray", vmin=0, vmax=255)
    ax.set_title(title)
    ax.axis("off")
plt.tight_layout()
plt.show()
 
for title, img in gradients:
    print(f"{title}: unique values = {len(np.unique(img))}")
```
