                            Exercise 3: for and while Loops

```python
pixels = [20, 130, 255]

for intensity in pixels:
    if intensity >= 128:
        print(intensity, "Bright pixel")
    else:
        print(intensity, "Dark pixel")


count = 0

while count < 3:
    print(count)
    count = count + 1
```

                                    Exercise 4: Functions

```python
def is_bright(pixel, threshold=128):

    if pixel == threshold:
        return True

    if pixel != threshold:
        return False


result1=is_bright(127)
result2= is_bright(128)
result3=is_bright(200,210)

print(result1)
print(result2)
print(result3)
```

                            Exercise 6: Dictionaries and Sets
